```python
import datasets
ds = datasets.load_dataset("openai/gsm8k", split="train", name="main")


import re
from collections import defaultdict

# Extract all final answers and group by their format
answer_formats = defaultdict(list)

# Pattern to extract the final answer after ####
final_answer_pattern = r'####\s*(.+)$'

for idx, answer in enumerate(ds['answer']):
    match = re.search(final_answer_pattern, answer, re.MULTILINE)
    if match:
        final_ans = match.group(1).strip()
        
        # Categorize the answer format
        if re.match(r'^-?\d+$', final_ans):
            category = 'integer'
        elif re.match(r'^-?\d+\.\d+$', final_ans):
            category = 'decimal'
        elif re.match(r'^-?\d+/\d+$', final_ans):
            category = 'fraction'
        elif re.match(r'^\$-?\d+(?:\.\d+)?$', final_ans):
            category = 'dollar_amount'
        elif re.match(r'^\$-?\d+/\d+$', final_ans):
            category = 'dollar_fraction'
        elif re.match(r'^-?\d+(?:\.\d+)?%$', final_ans):
            category = 'percentage'
        elif re.match(r'^-?\d+\s*/\s*\d+$', final_ans):
            category = 'fraction_with_spaces'
        elif ',' in final_ans and re.match(r'^-?\d{1,3}(,\d{3})*(?:\.\d+)?$', final_ans):
            category = 'number_with_commas'
        elif re.match(r'^\$-?\d{1,3}(,\d{3})*(?:\.\d+)?$', final_ans):
            category = 'dollar_with_commas'
        else:
            category = 'other'
        
        answer_formats[category].append({
            'index': idx,
            'final_answer': final_ans,
            'full_answer': answer
        })

# Print summary statistics
print("Answer Format Distribution:")
print("=" * 60)
for category in sorted(answer_formats.keys(), key=lambda x: len(answer_formats[x]), reverse=True):
    count = len(answer_formats[category])
    percentage = count / len(ds) * 100
    print(f"{category:25s}: {count:6d} ({percentage:5.2f}%)")

print("\n" + "=" * 60)
print(f"Total samples analyzed: {len(ds)}")

# Show examples from each category
print("\n\nExamples from each category:")
print("=" * 60)
for category in sorted(answer_formats.keys()):
    print(f"\n{category.upper()}:")
    # Show up to 3 unique examples
    unique_examples = []
    seen = set()
    for item in answer_formats[category]:
        if item['final_answer'] not in seen and len(unique_examples) < 3:
            unique_examples.append(item)
            seen.add(item['final_answer'])
    
    for example in unique_examples:
        print(f"  - '{example['final_answer']}'")
        print(f"    Question: {ds['question'][example['index']][:80]}...")
        print()
```

```python
import re

def reformat_answer(answer: str) -> str:
    """Replace '#### <ans>' with '<answer>{ans}</answer>' format."""
    pattern = r'####\s*(.+)$'
    match = re.search(pattern, answer, re.MULTILINE)
    
    if match:
        final_ans = match.group(1).strip()
        # remove the #### line and replace with <answer> tags
        reformatted = re.sub(pattern, f'<answer>{final_ans}</answer>', answer, flags=re.MULTILINE)
        return reformatted
    
    return answer

# Create the SFT dataset
sft_data = []
for idx in range(len(ds)):
    question = ds['question'][idx]
    answer = ds['answer'][idx]
    
    reformatted_answer = reformat_answer(answer)
    
    sft_data.append({
        'input': question,
        'output': reformatted_answer
    })

# Show some examples
print("SFT Dataset Examples:")
print("=" * 80)
for i in range(3):
    print(f"\n--- Example {i+1} ---")
    print(f"INPUT:\n{sft_data[i]['input']}")
    print(f"\nOUTPUT:\n{sft_data[i]['output']}")
    print()

print(f"\nTotal samples in SFT dataset: {len(sft_data)}")
```

```python
import re

def clean_calculator_annotations(text: str) -> str:
    """Remove <<a op b=c>> patterns, keeping just the result."""
    return re.sub(r'<<[^>]+>>', '', text)

# Create cleaned SFT dataset
sft_data_clean = []
for idx in range(len(ds)):
    question = ds['question'][idx]
    answer = ds['answer'][idx]
    
    # clean calculator annotations and reformat answer tag
    cleaned_answer = clean_calculator_annotations(answer)
    reformatted_answer = reformat_answer(cleaned_answer)
    
    sft_data_clean.append({
        'input': question,
        'output': reformatted_answer
    })

# Show before/after examples
print("Before/After Comparison:")
print("=" * 80)
for i in range(3):
    print(f"\n--- Example {i+1} ---")
    print(f"INPUT:\n{sft_data_clean[i]['input']}")
    print(f"\nORIGINAL OUTPUT:\n{ds['answer'][i]}")
    print(f"\nCLEANED OUTPUT:\n{sft_data_clean[i]['output']}")
    print("-" * 80)

print(f"\nTotal samples in cleaned SFT dataset: {len(sft_data_clean)}")
```

```python
import re
import json

SYSTEM_PROMPT = (
    "You are a helpful math assistant. Always provide your final numerical answer "
    "inside of the <answer>...</answer> tags, e.g.: <answer>42</answer>"
)

def clean_calculator_annotations(text: str) -> str:
    """Remove <<a op b=c>> patterns, keeping just the result."""
    return re.sub(r'<<[^>]+>>', '', text)

def reformat_to_answer_tags(answer: str) -> str:
    """Replace '#### <ans>' with '<answer>{ans}</answer>' format."""
    pattern = r'####\s*(.+)$'
    match = re.search(pattern, answer, re.MULTILINE)
    if match:
        final_ans = match.group(1).strip()
        return re.sub(pattern, f'<answer>{final_ans}</answer>', answer, flags=re.MULTILINE)
    return answer

def create_sft_sample(question: str, answer: str) -> dict:
    """Create a messages-formatted SFT sample."""
    cleaned = clean_calculator_annotations(answer)
    reformatted = reformat_to_answer_tags(cleaned)
    
    return {
        "messages": [
            {"role": "system", "content": SYSTEM_PROMPT},
            {"role": "user", "content": question},
            {"role": "assistant", "content": reformatted},
        ]
    }

# Generate the full SFT dataset in messages format
sft_messages_data = []
for idx in range(len(ds)):
    sample = create_sft_sample(ds['question'][idx], ds['answer'][idx])
    sft_messages_data.append(sample)

# Show examples
print("SFT Messages Format Examples:")
print("=" * 80)
for i in range(2):
    print(f"\n--- Example {i+1} ---")
    for msg in sft_messages_data[i]['messages']:
        role = msg['role'].upper()
        content = msg['content']
        if len(content) > 200:
            content = content[:200] + "..."
        print(f"[{role}]: {content}")
    print()

print(f"Total samples: {len(sft_messages_data)}")

# Preview the JSON structure
print("\n" + "=" * 80)
print("JSON structure preview:")
print(json.dumps(sft_messages_data[0], indent=2))
```
